🔴 Problem 0: Find Maximum Number

🧠 Problem

Given an integer array nums, return the maximum number in the array.  

nums = [3,7,2,9,5]  
output = 9

````python
nums = [3,7,2,9,5]  

def maxnum(nums):
    cn = nums[0]
    for i in nums:
        print("i: ",i)
        if i >= cn:
            cn = i
            print("cn: ",cn)
            
    return cn
        
        
maxnum(nums)
````

# 🔴 Problem 1: Two Sum (LeetCode #1)

nums = [2,7,11,15], target = 9  
output = [0,1]

````python
# Brute force

nums = [2,7,11,15]
target = 9

def twosum(nums, target):
    for i in range(len(nums)):
        for j in range(i+1, len(nums)):
            if nums[i] + nums[j] == target:
                return[i, j]
        
        
twosum(nums, target)
````

````python
# Optimized solution

nums = [2,7,11,15]
target = 9
d = {}


def twosum(nums, target, d):
    for i in range(len(nums)):
        num = nums[i]
        comp = target - num
        if comp in d:
            return [d[comp], i]
        else:
            d[num] = i
        
        
twosum(nums, target, d)
````

# 🔴 Problem 2: Contains Duplicate (LeetCode #217)
🧠 Problem

Given an integer array nums, return true if any value appears at least twice in the array, and return false if every element is distinct.  

🔁 Examples

nums = [1,2,3,1]  
output = True  

nums = [1,2,3,4]  
False  

````python
# Set

nums = [1,2,3,4]

def duplicate(nums):
    d = set()
    for i in nums:
        if i in d:
            return True
        else:
            d.add(i)
            
        return(False)
            
            
print(duplicate(nums))
````

````python
# Dictionary

nums = [1,2,3,4,1]

def duplicate(nums):
    d = {}
    for i in nums:
        if i in d:
            return(True)
        else:
            d[i]=i
            
    return False
    
            
            
print(duplicate(nums))
````

🔴 Problem 3: Valid Anagram (LeetCode #242)

🧠 Problem

Given two strings s and t, return True if t is an anagram of s, otherwise return False.

👉 An anagram = same characters, same frequency, different order allowed
```python
Case 1:

    s = "anagram"
    t = "nagaram"
    output = True

Case 2:

    s = "rat"
    t = "car"
    output = False
```

````python
s = "anagram"
t = "nagaram"

s2 = "rat"
t2 = "car"

def anagram(s, t):
    
    if len(s) != len(t):
        return False
    
    d = {}
    
    for i in s:
        # if i in d:
        #     d[i] += 1
        # else:
        #     d[i] = 1
        d[i] = d.get(i, 0) + 1
                
    for i in t:
        if i in d:
            d[i] -= 1
            if d[i] == 0:
                d.pop(i)
        else:
            return False
        
    return not d
    
print(anagram(s,t))
print(anagram(s2,t2))
````


````python
s = "A man, a plan, a canal: Panama"

def palindrome(s):
    left = 0
    right = len(s) - 1
    
    while left< right:
        # Skip non-alphanumeric (left)
        while left < right and not s[left].isalnum():
            left += 1
            
        # Skip non-alphanumeric (right)
        while left < right and not s[right].isalnum():
            right -= 1
        
        #compare lowercase
        if s[left].lower() != s[right].lower():
            return False
        
        left += 1
        right -= 1
        
    return True
````

🔴 Problem 5: Merge Sorted Array (LeetCode #88)

🧠 Problem

You are given two sorted arrays nums1 and nums2.

Merge them into one sorted array.
```python
Case1:
    nums1 = [1,2,3,0,0,0]
    m = 3

    nums2 = [2,5,6]
    n = 3

    output = [1,2,2,3,5,6]
```
⚠️ IMPORTANT  
nums1 has extra space at end  
You must modify nums1 in-place  
⚔️ YOUR TASK  

👉 Step 1: Explain logic

(Hint: two pointers from end)

````python
nums1 = [1,2,3,0,0,0]
m = 3

nums2 = [2,5,6]
n = 3

def merge(nums1, m, nums2, n):
    p1 = m - 1
    p2 = n - 1
    p = m + n - 1
    
    while p2 >= 0:
        if p1 >= 0 and nums1[p1] > nums2[p2]:
            nums1[p] = nums1[p1]
            p1 -= 1
        else:
            nums1[p] = nums2[p2]
            p2 -= 1
        p -= 1
    print(nums1)  

merge(nums1, m, nums2, n)
````

🔴 Problem 6: Best Time to Buy and Sell Stock (LeetCode #121)

🧠 Problem

You are given an array prices.

👉 You want to:

buy one day  
sell later

👉 Maximize profit

🔁 Example  
prices = [7,1,5,3,6,4]  
output = 5

👉 Buy at 1, sell at 6

````python
prices = [7,1,5,3,6,4]

def stock(prices):
    min_price = prices[0]
    max_profit = 0
    
    for i in range(len(prices)):
        if prices[i] < min_price:
            min_price = prices[i]
        else:
            profit = prices[i] - min_price
            if  profit > max_profit:
                max_profit = profit
        
    return max_profit

print(stock(prices))
````
